```python
import pandas as pd
from scipy.stats import chisquare, ks_2samp
import numpy as np
from sklearn.model_selection import train_test_split

df = pd.read_csv('../data/telco_data.csv')
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

monthly_charges = df["MonthlyCharges"]

s1 = monthly_charges[2000:3000]
s2 = monthly_charges[1000:2000]


import matplotlib.pyplot as plt

p_values = []
shifts = np.linspace(0,1,100)

for shift in shifts:
    result = ks_2samp(s1, s2 + shift)
    p_values.append(result.pvalue)

plt.plot(shifts, p_values, marker='o')
plt.xlabel("Shift applied to s2")
plt.ylabel("KS test p-value")
plt.axhline(y=0.05, color='r', linestyle='--', label='p=0.05 threshold')
plt.legend()
plt.show()


print(object.pvalue)
print(object_s.pvalue)
```
